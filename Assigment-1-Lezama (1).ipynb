```python
import pandas as pd
import numpy as np

df = pd.read_csv("/Users/christinelezama/Downloads/DT_coding_sample_training.csv")

LABEL_COL = "malicious"
```

Function 1

```python
def createBranches(att):
    branches = df[att].unique()

    return list(branches)
```

Function 2

```python
def divideByLabel(path):
    rows = df

    for branch in path:
        att = branch[0]
        value = branch[1]

        rows = rows[rows[att] == value]

    breakdown = {}

    labels = df[LABEL_COL].unique()

    for label in labels:
        count = len(rows[rows[LABEL_COL] == label])

        breakdown[label] = count

    return breakdown
```

Function 3

```python
def calcEntropy(aDict):
    total = sum(aDict.values())

    if total == 0:
        return 0.0

    entropy = 0.0

    for label in aDict:
        count = aDict[label]

        if count > 0:
            p = count / total

            entropy = entropy - (p * np.log2(p))

    return entropy
```

Function 4

```python
def findSplitAtt(branch, attList):
    parentRows = df

    for pathBranch in branch:
        att = pathBranch[0]
        value = pathBranch[1]

        parentRows = parentRows[
            parentRows[att] == value
        ]

    parentTotal = len(parentRows)
    parentBreakdown = divideByLabel(branch)
    parentEntropy = calcEntropy(
        parentBreakdown
    )

    bestAtt = None
    bestGain = -1

    for att in attList:

        weightedEntropy = 0.0
        branches = createBranches(att)

        for branchValue in branches:

            newPath = branch + [
                (att, branchValue)
            ]

            childRows = df

            for pathBranch in newPath:
                pathAtt = pathBranch[0]
                pathValue = pathBranch[1]

                childRows = childRows[
                    childRows[pathAtt] == pathValue
                ]

            weight = (
                len(childRows) / parentTotal
            )
            childBreakdown = divideByLabel(
                newPath
            )
            childEntropy = calcEntropy(
                childBreakdown
            )
            weightedEntropy = (
                weightedEntropy
                + weight * childEntropy
            )

        gain = (
            parentEntropy
            - weightedEntropy
        )

        if gain > bestGain:
            bestGain = gain
            bestAtt = att

    return bestAtt
```

Function 5

```python
def checkStopping(branch, attList):
    rows = df

    for pathBranch in branch:
        att = pathBranch[0]
        value = pathBranch[1]
        rows = rows[
            rows[att] == value
        ]

    if len(rows) == 0:
        return "no_data"
        
    labels = rows[LABEL_COL].unique()

    if len(labels) == 1:
        return "pure"
    if len(attList) == 0:
        return "no_attributes"

    return "continue"
```

Function 6

```python
def main(df):
    allAttributes = []

    for column in df.columns:
        if column != LABEL_COL:
            allAttributes.append(column)

    completePaths = []

    workList = [
        [[], allAttributes]
    ]

    position = 0
    while position < len(workList):

        current = workList[position]
        position = position + 1
        path = current[0]
        attList = current[1]
        stopping = checkStopping(
            path,
            attList
        )

        if stopping == "pure":

            breakdown = divideByLabel(path)
            leafLabel = None
            bestCount = -1

            for label in breakdown:

                if breakdown[label] > bestCount:
                    bestCount = breakdown[label]
                    leafLabel = label

            completePaths.append(
                path + [leafLabel]
            )

        elif stopping == "no_attributes":

            breakdown = divideByLabel(path)
            leafLabel = None
            bestCount = -1

            for label in breakdown:

                if breakdown[label] > bestCount:
                    bestCount = breakdown[label]
                    leafLabel = label

            completePaths.append(
                path + [leafLabel]
            )

        elif stopping == "no_data":

            parentPath = path[
                0:len(path) - 1
            ]
            breakdown = divideByLabel(
                parentPath
            )
            leafLabel = None
            bestCount = -1

            for label in breakdown:

                if breakdown[label] > bestCount:
                    bestCount = breakdown[label]
                    leafLabel = label

            completePaths.append(
                path + [leafLabel]
            )

        else:

            splitAtt = findSplitAtt(
                path,
                attList
            )

            remainingAtts = []

            for att in attList:

                if att != splitAtt:
                    remainingAtts.append(att)

            branches = createBranches(
                splitAtt
            )

            for branchValue in branches:

                newPath = path + [
                    (splitAtt, branchValue)
                ]

                workList.append(
                    [
                        newPath,
                        remainingAtts
                    ]
                )

    return completePaths
```

```python
result = main(df)

for path in result:
    print(path)
```
