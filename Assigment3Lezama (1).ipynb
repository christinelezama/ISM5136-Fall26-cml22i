# K-means Clustering Implementation

```python
import numpy as np
import pandas as pd

df = pd.read_csv(
    "/Users/christinelezama/Downloads/k-means_question_sample_data.csv",
    header=None
)

data = df.to_numpy()

print("Rows and columns:", data.shape)
```

```python
def standardizer(arr):
    averages = arr.mean(axis=0)
    spreads = arr.std(axis=0)

    spreads[spreads == 0] = 1

    return (arr - averages) / spreads


def calculateDistance(obj1, obj2):
    return np.sqrt(np.sum((obj1 - obj2) ** 2))


def findClosestCentroid(distArray):
    return np.argmin(distArray)


def calculateCentroids(assignmentArr):
    return assignmentArr.mean(axis=0)
```

```python
def main(arr, k):
    scaledData = standardizer(arr)

    if k < 1 or k > len(scaledData):
        raise ValueError("Choose k between 1 and the number of rows.")

    startingRows = np.random.choice(len(scaledData), size=k, replace=False)
    centroids = scaledData[startingRows].copy()
    previousLabels = np.full(len(scaledData), -1)

    for iteration in range(1000):
        labels = np.zeros(len(scaledData), dtype=int)

        for row in range(len(scaledData)):
            distances = []
            for center in centroids:
                distances.append(calculateDistance(scaledData[row], center))
            labels[row] = findClosestCentroid(distances)

        if np.array_equal(labels, previousLabels):
            return labels, centroids

        # Update each centroid using its assigned rows.
        for cluster in range(k):
            members = scaledData[labels == cluster]
            if len(members) > 0:
                centroids[cluster] = calculateCentroids(members)

        previousLabels = labels

    raise RuntimeError("Cluster assignments did not stabilize.")
```

```python
np.random.seed(42)

k = 3
labels, centroids = main(data, k)

print("First 10 cluster labels:", labels[:10])
print("Cluster sizes:", np.bincount(labels, minlength=k))
print("Final centroids:")
print(np.round(centroids, 3))
```

I standardized the five columns before calculating distances so they were on a comparable scale. For the sample run, I used three clusters. The algorithm kept assigning rows to the closest centroid and recalculating the centroids until the assignments stopped changing. The final clusters had 33, 34, and 33 rows.
